```python
# Cell 3: Génération de Données Globales

import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

def generate_server_data(days=30):
    """Génère des métriques serveur (CPU, RAM, Réseau)."""
    date_range = pd.date_range(start=datetime.now() - timedelta(days=days), periods=days*24, freq='H')
    data = []
    for date in date_range:
        # Simulation de patterns quotidiens (pics en journée)
        hour = date.hour
        # Formule sinusoïdale pour simuler les pics d'activité
        base_load = 30 + 20 * np.sin((hour - 6) * np.pi / 12) 
        
        cpu = base_load + np.random.normal(0, 5)
        ram = 40 + (cpu * 0.5) + np.random.normal(0, 2)
        latency = 20 + (cpu * 1.5) + np.random.normal(0, 10)
        latency = max(5, latency) # Latence minimale de 5ms
        
        data.append({
            "timestamp": date,
            "cpu_usage": round(cpu, 2),
            "ram_usage": round(ram, 2),
            "latency_ms": round(latency, 2),
            "request_count": int(cpu * 50)
        })
    return pd.DataFrame(data)

def generate_user_logs(n=100):
    """Génère des logs utilisateurs (Agents IA)."""
    agents = ['Analyste', 'Optimiseur', 'Prédicteur', 'Critique']
    actions = ['Data processing', 'Optimisation', 'Forecast update', 'Audit']
    statuses = ['success', 'warning', 'error']
    
    data = []
    for _ in range(n):
        data.append({
            "timestamp": datetime.now() - timedelta(minutes=random.randint(1, 1000)),
            "agent": random.choice(agents),
            "action": random.choice(actions),
            "status": random.choices(statuses, weights=[0.8, 0.15, 0.05])[0]
        })
    return pd.DataFrame(data).sort_values("timestamp")

# Génération effective des datasets
df_server = generate_server_data(days=15) # 15 jours de données
df_logs = generate_user_logs(n=200)

print("📈 Données Serveur générées (15 jours, fréquence horaire) :")
print(df_server.head())
print("\n📝 Logs Utilisateurs générés (200 entrées) :")
print(df_logs.head())
```
