```python
# Cellule 1 : Données d'Activité
import numpy as np
import pandas as pd

users = [101, 102, 103, 104, 105]
# Logs d'activité (Heure, Tentatives login, Echecs login)
logs = [
    (101, 8, 0), # User 101, 8h, 0 erreur
    (101, 14, 0),
    (102, 23, 1), # 1 erreur -> suspect
    (102, 23, 2), # 2 erreurs à la suite -> critique
    (103, 2, 0),
    (104, 15, 5), # Beaucoup d'erreurs -> Très critique
    (105, 10, 0),
]

df_logs = pd.DataFrame(logs, columns=['user_id', 'hour', 'failed_attempts'])
df_logs
```

```python
# Cellule 2 : Calcul du Score de Risque (Agent Critique)
def calculate_risk_score(row):
    score = 0
    
    # Règle 1 : Heures nocturnes (23h - 5h) augmentent le risque
    if row['hour'] < 5 or row['hour'] > 22:
        score += 20
        
    # Règle 2 : Échecs de login
    score += (row['failed_attempts'] * 30)
    
    return score

df_logs['risk_score'] = df_logs.apply(calculate_risk_score, axis=1)

df_logs['risk_level'] = pd.cut(df_logs['risk_score'], 
                                bins=[-1, 20, 50, 100], 
                                labels=['Faible', 'Moyen', 'Critique'])

df_logs
```
